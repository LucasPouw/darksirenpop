```python
import pandas as pd
from popsummary import PopulationResult
import matplotlib.pyplot as plt

from darksirenpop.utilities.utils import make_nice_plots
from darksirenpop.utilities.default_globals import *

make_nice_plots()
```

```python
result = PopulationResult(fname=LVK_HYPERPOSTERIOR_PATH)
df = pd.DataFrame(result.get_hyperparameter_samples(), columns=result.get_metadata("hyperparameters"))

# Relevant hyperparameters
mass_hyperparam = ['alpha_1', 'alpha_2', 'break_mass', 'mpp_1', 'sigpp_1', 'mpp_2', 'sigpp_2', 'mlow_1', 'delta_m_1', 'lam_0', 'lam_1', 'mmax',
                    'beta', 'mlow_2', 'delta_m_2']
spin_hyperparams = ['mu_chi', 'sigma_chi', 'mu_spin', 'sigma_spin', 'xi_spin']
z_hyperparams = ['z_peak', 'gamma', 'kappa']
```

```python
print(np.median(df['gamma']), np.median(df['gamma']) - np.quantile(df['gamma'], 0.16), np.quantile(df['gamma'], 0.84) - np.median(df['gamma']))

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(24, 6))
ax1.hist(df['gamma'], bins=100, density=True)
ax1.set_xlabel(r'$\gamma$')
ax1.set_ylabel('Probability density')
ax2.hist(df['kappa'], bins=100, density=True)
ax2.set_xlabel(r'$\kappa$')

ax3.hist(df['z_peak'], bins=100, density=True)
ax3.set_xlabel(r'$z_{\rm peak}$')

plt.tight_layout()
plt.show()
```
